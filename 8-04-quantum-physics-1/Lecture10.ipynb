## <u>Stationary states</u>

Now, we return to the Schrödinger Equation

$$
i\hbar\pd{\Psi}{t}= \left(\frac{h^2}{2m} + V(x)\right)\Psi
$$

And ask how can we solve it?

Answer: Separation of variables!

## <u>Assume $\Psi(x, t) = g(t)\psi(t)$</u>

Then we plug it into the Schrödinger Equation

$$
\begin{align*}
i \hbar \pd{\Psi}{t} &= 
    \left(
        \frac{\hbar^2}{2m}
        \frac{\partial^2}{\partial x^2} + V(x) 
    \right) \Psi(x, t)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow 
i \hbar \pdo{t}{g(t)\psi(x)} &=
    \left(
        \frac{\hbar^2}{2m}
        \frac{\partial^2}{\partial x^2} 
        +
        V(x)
    \right) g(t)\psi(x)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow
i \hbar \psi(x)\do{t}{g(t)} &=
    \left(
        \frac{\hbar^2}{2m}
        \frac{\partial^2}{\partial x^2}
        g(t)\psi(x)
        +
        V(x)g(t)\psi(x)
    \right)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow
i \hbar \psi(x)\do{t}{g(t)} &=
    \left(
        \frac{\hbar^2}{2m}
        g(t)
        \frac{\d^2}{\d x^2}
        \psi(x)
        +
        V(x)g(t)\psi(x)
    \right)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow
i \hbar \psi(x)\do{t}{g(t)} &=
    g(t)
    \left(
        \frac{\hbar^2}{2m}
        \frac{\d^2}{\d x^2}
        \psi(x)
        +
        V(x)\psi(x)
    \right)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow
i \hbar \psi(x)\do{t}{g(t)} &=
    g(t)
    \left(
        \frac{\hbar^2}{2m}
        \frac{\d^2}{\d x^2}
        +
        V(x)
    \right)
    \psi(x)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow
i \hbar \psi(x)\do{t}{g(t)} &=
    g(t)
    \hat{H}
    \psi(x)\\\\
\end{align*}
$$

$$
\begin{align*}
\Rightarrow
\frac{i \hbar}{g(t)} \do{t}{g(t)} &=
    \frac{1}{\psi(x)}
    \hat{H}
    \psi(x)\\\\
\end{align*}
$$

**But Wait**

1. Left hand side is a function of $t$
2. Right hand side is a function of $x$

$$
\begin{align*}
\Rightarrow
\overbrace{\frac{i \hbar}{g(t)} \do{t}{g(t)}}^\text{function of $t$}\; 
&=
    \overbrace{\frac{1}{\psi(x)}
    \hat{H}
    \psi(x)}^\text{function of $x$}
\end{align*}
$$

Now note that the left hand side is a function of time alone, while the right hand side is a function of space alone. How can this be? By varying the value of time I could change the left hand side, but it still has to be equal to the value of $x$ for all $x$. The only way for this to work is if both the left and right hand sides are _constant_! We will call this constant $E$, and we then have two ordinary differential equations,

**Assume that both are equal to a constant $E$**

$$
\begin{align*}
\Rightarrow
\frac{i \hbar}{g(t)} \do{t}{g(t)} &=
    \frac{1}{\psi(x)}
    \hat{H}
    \psi(x) = E\\\\
\end{align*}
$$

**Now we get a system of two uncoupled differential equations**

$$
\begin{align}
\frac{i \hbar}{g(t)} \do{t}{g(t)} &= E \\\\
\frac{1}{\psi(x)}\hat{H}\psi(x) &= E
\end{align}
$$

**The first is easy to solve**

$$
\frac{i\hbar}{g(t)}\do{t}{g(t)} = E
$$

$$
\do{t}{g(t)} = \frac{-iE}{\hbar}g(t)
$$

$$
g(t) = Ce^{\frac{-iE}{\hbar}t}
$$

**The second is the Time-Independent Schrödinger Equation**

$$
\begin{align}
\frac{1}{\psi(x)}\hat{H}\psi(x) &= E \\\\ \Rightarrow \hat{H}\psi &= E\psi
\end{align}
$$

So we can write the general form of a stationary state solution

$$
\Psi(x, t) = \psi(x)e^{\frac{-iEt}{\hbar}} \quad \text{where}\; E \in \mathbb{R} \; \text{and} \; \hat{H}\psi = E\psi
$$

## Notes

### Note 1) $E$ must be real, since otherwise the state won't normalize

$$
\begin{align}
1 &= \int{\d{x} \Psi^*(x, t) \Psi(x, t)}\\\\
&= \int{\d{x} \psi^*(x)e^{\frac{iE^*t}{\hbar}} e^{\frac{-iEt}{\hbar}}\psi(x)}\\\\
&= e^{\frac{i(E^* - E)t}{\hbar}}\int{\d{x} \psi^*(x)\psi(x)}\\\\
&= \underbrace{e^{\frac{2i\operatorname{Im}(E)t}{\hbar}}}_{\text{varies for non-real E!}}\int{\d{x} \psi^*(x)\psi(x)}
\end{align}
$$

**Also note this implies that, since $E\in\mathbb{R}$, $\psi$ is normalized as well**

It's worth showing that $E$ is real explicitly, even though we know it should be real assuming that $H$ is a Hermitian operator

### Note 2) $E$ is the energy of the state

The energy of the state is well-defined since $\psi$ is an eigenstate of $\hat{H}$, with eigenvalue $E$

### Note 3) Expectation Values on Stationary States Don't Change

For any observable $Q$ that does not explicitly involve time, the expectation value of that operator $\langle Q \rangle$ does not change in time for a stationary state

$$
\begin{align}
\langle Q \rangle &= \int{\d{x} \Psi^*(x, t)Q\Psi(x,t) }\\\\
&= \int{\d{x} \psi^*(x) e^{\frac{iEt}{\hbar}} e^{\frac{-iEt}{\hbar}}Q\psi(x)}\quad\text{Assuming $Q$ does not act on functions of $t$}\\\\
&= \int{\d{x} \psi^*(x) Q \psi(x)} \quad \bigg\rbrace \; \text{not a function of $t$}
\end{align}
$$


### Note 4) Superposition of Stationary States is NOT Stationary

Assuming the statese have different energies.

Firstly, the state does not obviously factorize into $g(t)\psi(x)$

$$
\Psi(x, t) = c_1\psi_1(x)e^{\frac{-iE_1t}{\hbar}} + c_2\psi_2(x)e^{\frac{-iE_2t}{\hbar}}
$$

Furthermore, we can see that expectation values for $Q$ and $\Psi$ _do_ change with time

$$
\begin{align}
\langle Q \rangle &= \int{\d{x} \Psi^*\Psi}\\\\
&= \int{\d{x} \left(c_1^*\psi^*_1(x)e^{\frac{iE_1t}{\hbar}} + c^*_2\psi^*_2(x)e^{\frac{iE_2t}{\hbar}}\right)}Q\left(c_1\psi_1(x)e^{\frac{-iE_1t}{\hbar}} + c_2\psi_2(x)e^{\frac{-iE_2t}{\hbar}}\right)\\\\
&= \int{\d{x} c_1^*c_1\psi_1^*Q\psi_1 + c_2^*c_2\psi_2^*Q\psi_2 + c_1^*c_2e^{-i(E_1-E_2)t/\hbar}\psi_1^*Q\psi_2 + c_2^*c_1e^{-i(E_2-E_1)t/\hbar}\psi_2^*Q\psi_1}\\\\
&= |c_1|^2\langle Q\rangle_{\psi_1} + |c_2|^2\langle Q\rangle_{\psi_2} + \underbrace{2\operatorname{Re}\left[c_1^*c_2e^{-i(E_1-E_2)t/\hbar}\int{\d{x} \psi_1^*Q\psi_2}\right]}_{\text{varies with time!}}
\end{align}
$$

## Solving For Energy Eigenstates

Since the Time-Independent Schrödinger Equation is the eigenvalue equation for a Hermitian operator $\hat{H}$, we observe that:

1. The allowed values of $E$ are not arbitrary. There are definite, possibly discrete, energy levels for stationary states
2. The energy eigenstates $\psi_E$ form an orthonormal set of vectors. That is

$$
E_1 \neq E_2 \Rightarrow \int{\d{x} \psi^*_{E_1}\psi_{E_2}} = 0
$$

If we have no degeneracy in our Hamiltonian (all energy eigenstates have distinct energies, so there are no energy eigenstates that share an energy level), the converse becomes true as well, and the eigenfunctions would form an orthonormal basis for all possible quantum states.

In this case, we can solve the general Schrödinger Equation by projecting our initial state $\Psi(x, 0)$ onto the energy eigenstates, attaching the $e^{-iEt/\hbar}$ factor, and recombining them:

$$
\Psi(x, t) = \sum_{n=1}^\infty{c_n\psi_n(x)e^{-iE_n t/\hbar}} \quad \text{where} \quad c_n = \int{\d{x} \psi^*_n(x)\Psi(x, 0)}
$$

## Free particle on a circle

### Now consider a particle constrained to a circle of circumference $L$

1. The topology of this condition implies that $\psi(x + L) = \psi(x)$ since $x + L$ is essentially the same point as $x$.


2. Assume the particle is free $V = 0$

Now, the Time-Independent Schrödinger Equation looks like

$$
-\frac{\hbar^2}{2m}\frac{\d{}^2\psi}{\d{x}^2} = E\psi
$$

First we prove that $E$ is positive

$$
-\frac{\hbar^2}{2m}\frac{\d{}^2\psi}{\d{x}^2} = E\psi
$$

$$
\Rightarrow
-\frac{\hbar^2}{2m}\int{\d{x}\psi^*(x)\frac{\d{}^2\psi}{\d{x}^2}}= \int{\d{x} \psi^*E\psi}
$$

$$
\Rightarrow
-\frac{\hbar^2}{2m}\int{\d{x}\psi^*(x)\frac{\d{}^2\psi}{\d{x}^2}}= E \int{\d{x}\psi^*\psi}
$$

$$
\Rightarrow
-\frac{\hbar^2}{2m}\int{\d{x}\left[\do{x}{\left(\psi^*\td{\psi}{x}\right)} - \td{\psi^*}{x}\td{\psi}{x}\right]}= E
$$

$$
\Rightarrow
-\frac{\hbar^2}{2m}\left(\left[\psi^*\td{\psi}{x}\right]^L_0 - \int{\d{x} \td{\psi^*}{x}\td{\psi}{x}}\right) = E
$$

$$
\Rightarrow
0 \leq \frac{\hbar^2}{2m}\int{\d{x} \td{\psi^*}{x}\td{\psi}{x}} = E
$$

Next we solve the differential equation by making the substitution $k^2 = \frac{2mE}{\hbar^2}$

$$
-\frac{\hbar^2}{2m}\frac{\d{}^2\psi}{\d{x}^2} = E\psi
$$

$$
\frac{\d{}^2\psi}{\d{x}^2} = -\frac{2m E}{\hbar^2}\psi = -k^2\psi
$$

This happens to be simple harmonic motion, so it is solved by

$$
\begin{align}
\psi(x) &= \sin{kx}\\
\psi(x) &= \cos{kx}\\
\psi(x) &= e^{ikx}
\end{align}
$$

Focusing on $e^{ikx}$, we can see that if $x + L$ is the same as $x$, we must have

$$
\begin{align}
\psi(x + L) = \psi(x) = e^{ik(x + L)} &= e^{ikx} \\\\ \Rightarrow e^{ikL} &= 1 \\\\ \Rightarrow kL &= 2\pi n \\\\ \Rightarrow k &= \frac{2\pi n}{L} \quad \forall n \in \mathbb{Z}
\end{align}
$$

Thus, we have discrete energy levels given by 

$$
k^2 = \frac{4\pi^2 n^2}{L^2} = \frac{2mE}{\hbar^2} \Rightarrow E = \frac{2\pi^2n^2\hbar^2}{mL^2}
$$
