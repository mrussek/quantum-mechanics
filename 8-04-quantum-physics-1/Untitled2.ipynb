1. **Gaussians and uncertainty product saturation**

   Consider the gaussian wave function

   $$
   \psi(x) = N\exp{\left(-\frac{1}{2}\frac{x^2}{a^2}\right)}
   $$
   
   Where $N\in\mathbb{R}$ and $a$ is a real positive constant with units of length. Also consider the integral identities
   
   $$
   \int_{-\infty}^\infty{dxe^{-\alpha x^2 + \beta x}} = \sqrt{\frac{\pi}{\alpha}}\exp{\left(\frac{\beta^2}{4\alpha}\right)} \quad \operatorname{Re}(\alpha) > 0\\\\
   \int_{-\infty}^\infty{dx\;x^2 e^{-\alpha x^2}} = \frac{1}{2\alpha}\int_{-\infty}^\infty{dx e^{-\alpha x^2}}
   $$
   

   (a) Use the position space wavefunction (1) to calculate $\Delta x$ and $\Delta p$. Confirm that your answer saturates the Heisenberg uncertainty product
   
   $$
   (\Delta x)^2 = \langle x^2 \rangle - (\langle x \rangle)^2
   $$
   
   $$
   \Delta x \Delta p \ge \frac{\hbar}{2}
   $$
   
   First, note that the expectation value of $x$ is 0
   
   $$
   \begin{align}
   \langle x \rangle &= \int_{-\infty}^\infty{dx\,x | \psi^2(x) | }\\\\ &= \int_{-\infty}^\infty{dx\,x N^2 e^{\frac{-x^2}{a^2}}}\\\\
   &= N^2 \int_{-\infty}^\infty{dx\, x e^{\frac{-x^2}{a^2}}}\\\\
   &= N^2 \left[ \frac{-a^2}{2} e^{\frac{-x^2}{a^2}} \right]_{-\infty}^\infty\\\\
   &= 0
   \end{align}
   $$
   
   So then the uncertainty $\Delta x$ is just
   
   $$
   \Delta x = \sqrt{\langle x^2 \rangle}
   $$
   
   We can calculate
   
   $$
   \begin{align}
   \langle x^2 \rangle &= \int_{-\infty}^\infty{dx |\psi(x)|^2 x^2}\\\\
   &= \int_{-\infty}^\infty{dx  N^2 e^{\frac{-x^2}{a^2}} x^2}\\\\
   &= \frac{(Na)^2}{2}\int_{-\infty}^\infty{dx  e^{\frac{-x^2}{a^2}}}\\\\
   &= \frac{a^2}{2}
   \end{align}
   $$
   
   Then, similarly for $\Delta p$
   
   $$
   \begin{align}
   \langle p \rangle &= \int_{-\infty}^\infty{dx\, \psi^*(x)\hat{p}\psi(x)}\\\\
   &= \int_{-\infty}^\infty{dx\, \psi^*(x)(-i\hbar)\frac{d}{dx}\psi(x)}\\\\
   &= \int_{-\infty}^\infty{dx\, \psi^*(x)(-i\hbar)\frac{d}{dx}\psi(x)}\\\\
   &= -i\hbar N^2\int_{-\infty}^\infty{dx\, e^{-\frac{x^2}{2a^2}}\frac{d}{dx}e^{-\frac{x^2}{2a^2}}}\\\\
   &= \frac{i\hbar N^2}{a^2}\int_{-\infty}^\infty{dx\, x e^{-\frac{x^2}{2a^2}}e^{-\frac{x^2}{2a^2}}}\\\\
   &= \frac{i\hbar N^2}{a^2}\int_{-\infty}^\infty{dx\, x e^{-\frac{x^2}{a^2}}} = 0
   \end{align}
   $$
   
   So we calculate
   
   $$
   \begin{align}
   \langle \hat{p}^2 \rangle &= \int_{-\infty}^\infty{dx\,\psi^*\hat{p}^2\psi}\\\\
   &= \int_{-\infty}^\infty{dx\,(\hat{p}\psi)^*\hat{p}\psi}\\\\
   &= \int_{-\infty}^\infty{dx\,(\hat{p}\psi)^*\hat{p}\psi}\\\\
   &= \frac{N^2\hbar^2}{a^4}\int_{-\infty}^\infty{dx\,x^2e^{\frac{-x^2}{a^2}}}\\\\
   &= \frac{\hbar^2}{2a^2}
   \end{align}
   $$
   
   And can thus verify that $\Delta x \Delta p = \frac{\hbar}{2}$
